```python
data = conn.execute("""
    WITH award_counts AS (
    SELECT 
            r.ocid,
            r.release_id,
            r.release_date,
            COUNT(a.award_id) AS award_count
    FROM stg_releases r
    LEFT JOIN stg_awards a
    ON r.release_id = a.release_id
    group by r.ocid, r.release_id, r.release_date,
    ),

    award_changes AS (
        SELECT *,
               LAG(award_count) OVER (
                   PARTITION BY ocid
                   ORDER BY release_date
               ) AS previous_award_count
        FROM award_counts
    )
    
    SELECT *
    FROM award_changes
    where award_count != previous_award_count
    
""").df()

data
```
