```python
import requests

print(requests.__version__)
```

```python
url = "https://www.find-tender.service.gov.uk/api/1.0/ocdsReleasePackages"

response = requests.get(
    url,
    timeout=10
)

print("Status:", response.status_code)
print("Content type:", response.headers.get("content-type"))
print(response.text[:500])
```

```python
data = response.json()

print(type(data))
print(data.keys())
```

```python
tag_counts = {}

for release in releases:
    for tag in release["tag"]:
        if tag not in tag_counts:
            tag_counts[tag] = 0

        tag_counts[tag] += 1

print(tag_counts)
```

```python
ocid_counts = {}

for release in releases:
    ocid = release["ocid"]

    if ocid not in ocid_counts:
        ocid_counts[ocid] = 0

    ocid_counts[ocid] += 1
```

```python
for ocid, count in ocid_counts.items():
    if count > 1:
        print(ocid, count)
```

```python
url = "https://www.find-tender.service.gov.uk/api/1.0/ocdsReleasePackages"

all_releases = []
page = 1
max_pages = 10

while url and page <= max_pages:

    print(f"Downloading page {page}...")

    response = requests.get(
        url,
        timeout=10
    )

    response.raise_for_status()

    page_data = response.json()

    all_releases.extend(page_data["releases"])

    url = page_data["links"].get("next")

    page += 1

print("Total releases:", len(all_releases))
```

```python
ocid_counts = {}

for release in all_releases:
    ocid = release["ocid"]

    if ocid not in ocid_counts:
        ocid_counts[ocid] = 0

    ocid_counts[ocid] += 1
```

```python
for ocid, count in ocid_counts.items():
    if count > 1:
        print(ocid, count)
```

```python
target_ocid = "ocds-h6vhtk-077cb7"

matching_releases = []

for release in all_releases:
    if release["ocid"] == target_ocid:
        matching_releases.append(release)

print("Matching releases:", len(matching_releases))
```

```python
for release in matching_releases:
    print("ID:", release["id"])
    print("Date:", release["date"])
    print("Tag:", release["tag"])
    print("Tender title:", release["tender"]["title"])
    print("-" * 50)
```

```python
matching_releases.sort(key=lambda x: x["date"])

for release in matching_releases:
    print("ID:", release["id"])
    print("Date:", release["date"])
    print("Tag:", release["tag"])
    print()
```

```python
for i in range(1, len(matching_releases)):

    previous = matching_releases[i - 1]["tender"]
    current = matching_releases[i]["tender"]

    print(
        matching_releases[i - 1]["id"],
        "->",
        matching_releases[i]["id"]
    )

    all_keys = set(previous.keys()) | set(current.keys())

    for key in all_keys:
        if previous.get(key) != current.get(key):
            print("Changed:", key)

    print("-" * 50)
```
